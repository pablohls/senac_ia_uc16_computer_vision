# Segmentação por Threshold — Folhas

Carrega uma imagem de `data/images/standardleaves` e usa um slider para alterar o valor do threshold e ver as mudanças em tempo real.

```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# Garante que os widgets renderizem inline
%matplotlib inline
```

```python
# Pasta com as imagens (caminho relativo à raiz do projeto)
IMAGES_DIR = Path("../data/images/standardleaves")

# Escolha a imagem a carregar
image_path = IMAGES_DIR / "33.jpg"

# OpenCV lê em BGR; convertemos para RGB (exibição) e para cinza (threshold)
bgr = cv2.imread(str(image_path))
if bgr is None:
    raise FileNotFoundError(f"Não foi possível carregar a imagem: {image_path.resolve()}")

rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)

print(f"Imagem carregada: {image_path.name}  shape={rgb.shape}")
```

```python
def show_threshold(threshold=200):
    """Aplica um threshold binário e mostra original + resultado lado a lado."""
    _, mask = cv2.threshold(gray, threshold, 255, cv2.THRESH_BINARY)

    fig, axes = plt.subplots(1, 2, figsize=(11, 5))
    axes[0].imshow(rgb)
    axes[0].set_title("Original")
    axes[0].axis("off")

    axes[1].imshow(mask, cmap="gray")
    axes[1].set_title(f"Threshold = {threshold}")
    axes[1].axis("off")

    plt.tight_layout()
    plt.show()


# Slider em tempo real (0–255). Mova para ver o resultado mudar.
interact(
    show_threshold,
    threshold=IntSlider(min=0, max=255, step=1, value=127, continuous_update=True),
);
```

```python
src = cv2.imread(image_path)
src_gray = cv2.cvtColor(src, cv2.COLOR_BGR2GRAY)

# Apply different threshold types
threshold_value = 180
max_value = 255

# Binary threshold
_, binary = cv2.threshold(src_gray, threshold_value, max_value, cv2.THRESH_BINARY)

# Binary inverted threshold
_, binary_inv = cv2.threshold(src_gray, threshold_value, max_value, cv2.THRESH_BINARY_INV)

# Truncate threshold
_, truncate = cv2.threshold(src_gray, threshold_value, max_value, cv2.THRESH_TRUNC)

# To zero threshold
_, to_zero = cv2.threshold(src_gray, threshold_value, max_value, cv2.THRESH_TOZERO)

# To zero inverted threshold
_, to_zero_inv = cv2.threshold(src_gray, threshold_value, max_value, cv2.THRESH_TOZERO_INV)

cv2.imshow("Original", src_gray)
cv2.imshow("Binary", binary)
cv2.imshow("Binary Inverted", binary_inv)
cv2.imshow("Truncate", truncate)
cv2.imshow("To Zero", to_zero)
cv2.waitKey(0)
```

```python
src = cv2.imread(image_path)
gray = cv2.cvtColor(src, cv2.COLOR_BGR2GRAY)

# Global threshold
_, global_thresh = cv2.threshold(gray, 180, 255, cv2.THRESH_BINARY)

# Adaptive threshold (mean)
adaptive_mean = cv2.adaptiveThreshold(gray, 255, 
                                      cv2.ADAPTIVE_THRESH_MEAN_C,
                                      cv2.THRESH_BINARY, 11, 2)

# Adaptive threshold (gaussian)
adaptive_gaussian = cv2.adaptiveThreshold(gray, 255,
                                          cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                          cv2.THRESH_BINARY, 11, 2)

cv2.imshow("Original", gray)
cv2.imshow("Global Threshold", global_thresh)
cv2.imshow("Adaptive Mean", adaptive_mean)
cv2.imshow("Adaptive Gaussian", adaptive_gaussian)
cv2.waitKey(0)
```
